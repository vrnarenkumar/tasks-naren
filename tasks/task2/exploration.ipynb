# Task 2 — Exploring `table_1.csv` and `table_2.csv`
This task is aiming to provide a ≈≈ of the column “Type”. 

The dataset is provided in two different tables ("table_1.csv" and "table_2.csv")  with unique identifier of column “ID”. Tip: This column (ID) can be used to match the two tables. 

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
t1 = pd.read_csv("../../docs/table_1.csv", sep=";")
print("table_1 shape:", t1.shape)
```

```python
t2 = pd.read_csv("../../docs/table_2.csv", sep=";")
print("table_2 shape:", t2.shape)
```

```python
t1 = t1.drop_duplicates()
t1
```

```python
t2 = t2.drop_duplicates()
t2
```

```python
merged = t1.merge(t2, on="ID")
print("merged shape:", merged.shape)
print("duplicated IDs remaining:", merged["ID"].duplicated().sum())
merged.head()
```

```python
listall = merged.isna().mean().sort_values(ascending=False)
listall
```

```python
numeric_cols = ["UKL", "CDx", "NKJUD", "LPI", "POUG", "TRE", "OIN"]
categorical_cols = ["GJAH", "ZIK", "HUI", "ERZ", "BJZHD", "BJKG", "ZUB", "VOL", "UIO", "VBNM"]

df = merged.copy()

df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
df[categorical_cols] = df[categorical_cols].fillna("missing")

df.isna().sum().sum()  # sanity check: should be 0 now
```

```python
from sklearn.preprocessing import LabelEncoder

encoded = df.copy()
for col in categorical_cols:
    encoded[col] = LabelEncoder().fit_transform(encoded[col])

X = encoded[numeric_cols + categorical_cols]
y = (encoded["Type"] == "y").astype(int)  # 1 = y, 0 = n

X.shape, y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train.shape, X_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    roc_auc_score, accuracy_score, precision_score, recall_score,
    f1_score, classification_report, roc_curve, confusion_matrix,
)

try:
    from xgboost import XGBClassifier
    HAS_XGB = True
except Exception as e:
    HAS_XGB = False
    print(f"xgboost unavailable, skipping: {e}")

models = {}

# Logistic Regression
models["Logistic Regression"] = LogisticRegression(class_weight="balanced", max_iter=1000)

# Support Vector Machine
models["Support Vector Machine"] = LinearSVC(class_weight="balanced", max_iter=5000)

# Decision Tree
models["Decision Tree"] = DecisionTreeClassifier(class_weight="balanced", random_state=42)

# Random Forest
models["Random Forest"] = RandomForestClassifier(class_weight="balanced", random_state=42)

# Gradient Boosting
models["Gradient Boosting"] = GradientBoostingClassifier(random_state=42)

# Naive Bayes
models["Naive Bayes"] = GaussianNB()

# K-Nearest Neighbor
models["K-Nearest Neighbor"] = KNeighborsClassifier()

# XGBoost
if HAS_XGB:
    neg, pos = y_train.value_counts()[0], y_train.value_counts()[1]
    models["XGBoost"] = XGBClassifier(scale_pos_weight=neg / pos, random_state=42, eval_metric="logloss")

models
```

```python
SCALE_MODELS = {"Logistic Regression", "Support Vector Machine", "K-Nearest Neighbor"}

accuracy = {}
precision = {}
recall = {}

for name, model in models.items():
    Xtr, Xte = (X_train_scaled, X_test_scaled) if name in SCALE_MODELS else (X_train, X_test)
    model.fit(Xtr, y_train)
    preds = model.predict(Xte)

    accuracy[name] = accuracy_score(y_test, preds)
    precision[name] = precision_score(y_test, preds)
    recall[name] = recall_score(y_test, preds)

df_model = pd.DataFrame(index=models.keys(), columns=["Accuracy", "Precision", "Recall"])
df_model["Accuracy"] = accuracy.values()
df_model["Precision"] = precision.values()
df_model["Recall"] = recall.values()

df_model
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import make_pipeline

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring = ["accuracy", "precision", "recall", "f1", "roc_auc"]

cv_results = {}
for name, model in models.items():
    estimator = make_pipeline(StandardScaler(), model) if name in SCALE_MODELS else model
    scores = cross_validate(estimator, X, y, cv=cv, scoring=scoring)
    cv_results[name] = {metric: scores[f"test_{metric}"].mean() for metric in scoring}
    cv_results[name].update({f"{metric}_std": scores[f"test_{metric}"].std() for metric in scoring})

cv_df = pd.DataFrame(cv_results).T[["accuracy", "precision", "recall", "f1", "roc_auc"]].round(3)
cv_df
```

## Summary

**Challenges found in the data**
- `table_1` and `table_2` each contained 370 duplicated `ID`s (exact duplicate rows, not conflicting values). Left unhandled, a random train/test split would let identical rows leak across the split, inflating scores. Fixed by deduplicating each table before merging — 4070 → 3700 rows per table, 0 duplicate `ID`s after the merge.

- `ZIK` is ~58% missing; the remaining columns are missing at most ~2.7%. Imputed numeric columns with the median and categorical columns with a `"missing"` placeholder category.

- `Type` is heavily imbalanced (~92% `y` / ~8% `n`), so `class_weight="balanced"` was used where supported, and evaluation relied on precision/recall/F1/ROC-AUC rather than accuracy alone.

- Column names (`UKL`, `GJAH`, `ZIK`, ...) are anonymized/cryptic, so features were categorized by dtype and cardinality (7 numeric, 10 low-cardinality categorical) rather than by domain meaning.

**Approach**
Compared 8 classifiers (Logistic Regression, SVM, Decision Tree, Random Forest, Gradient Boosting, Naive Bayes, KNN, XGBoost) on the same deduplicated, imputed, label-encoded data, using both a held-out stratified split and 5-fold stratified cross-validation for robustness.

**Result**: Random Forest and XGBoost were the strongest and most consistent performers (~0.99 accuracy/precision/F1 in cross-validation, recall of 1.0). Random Forest is the recommended final model — it matches XGBoost's performance while being simpler to explain, and its feature importances offer a check against trivial data leakage (no single feature dominated importance, meaning the classes are genuinely separable from a combination of features rather than one leaking column).
