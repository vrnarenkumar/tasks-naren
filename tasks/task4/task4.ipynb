
**Task 4:**

Write a function that takes as input two timestamps of the form 2017/05/13 12:00 and 
calculates their differences in hours. 

**Example: 2022/02/15 00:05 and 2022/02/15 01:00 would return 1 hour.**


```python
import datetime
```

```python
def hours_between(time):
    start, end = time

    start_time = start.split(' ')[-1]
    start_date = start.split(' ')[0]

    end_time = end.split(' ')[-1]
    end_date = end.split(' ')[0]

    fmt = "%Y/%m/%d %H:%M"
    t1 = datetime.datetime.strptime(start_date + ' ' + start_time, fmt)
    t2 = datetime.datetime.strptime(end_date + ' ' + end_time, fmt)

    diff_hours = abs((t2 - t1).total_seconds()) / 3600
    return round(diff_hours)
```

```python

test_cases = [
    ("2022/02/15 00:05", "2022/02/15 01:00"),
    ("2017/05/13 12:00", "2017/05/13 12:00"),
    ("2017/05/13 12:00", "2017/05/13 12:29"),
    ("2017/05/13 12:00", "2017/05/13 12:30"),
    ("2017/05/13 12:00", "2017/05/13 12:31"),
    ("2017/05/13 12:00", "2017/05/14 12:00"),
    ("2017/05/13 12:00", "2017/05/13 09:00"),
    ("2017/01/31 23:30", "2017/02/01 00:00"),
    ("2017/12/31 23:00", "2018/01/01 01:00"),
    ("2020/02/28 23:00", "2020/03/01 01:00"),
    ("2021/02/28 23:00", "2021/03/01 01:00"),
    ("2017/05/13 00:00", "2019/05/13 00:00"),
]
```

```python
for case in test_cases:
    print(f"{case[0]} -> {case[1]} = {hours_between(case)}h")
```
