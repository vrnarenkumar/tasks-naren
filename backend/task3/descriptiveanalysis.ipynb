## Finding 5 alternative parts based on DESCRIPTION

Approach: only compare parts that have a DESCRIPTION. Filter candidates to the same Rated Current first (an "alternative" fuse must be electrically equivalent), then rank the remaining candidates by DESCRIPTION using lexical search

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# check what the DEFAULT tokenizer does to numbers before touching anything
default_analyzer = TfidfVectorizer().build_analyzer()
test_values = ["6.3A", "3A", "0.5A", "2.5A", "12.5A"]

for value in test_values:
    print(value, "->", default_analyzer(value))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

parts = df[df['DESCRIPTION'].notna()].reset_index(drop=True)
print("Parts with a description:", len(parts))

vectorizer = TfidfVectorizer(token_pattern=r'(?u)\b\w[\w.]*\b')
vectors = vectorizer.fit_transform(parts['DESCRIPTION'])
print("Vector shape:", vectors.shape)
```

```python
similarity_matrix = cosine_similarity(vectors)

for i in range(0, len(similarity_matrix)):
    similarity_matrix[i][i] = -1

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
match_count = 0
checked_count = 0

for i in range(0, len(parts)):
    current_i = parts['Rated Current (A) - cleaned'][i]
    if pd.notna(current_i):
        best_match_index = similarity_matrix[i].argmax()
        current_best = parts['Rated Current (A) - cleaned'][best_match_index]
        checked_count = checked_count + 1
        if current_i == current_best:
            match_count = match_count + 1

agreement = match_count / checked_count * 100
print("Parts checked:", checked_count)
print("Top match shares Rated Current:", match_count, f"({agreement:.1f}%)")
```

**Validation: 85.7% (504/588) of parts have a top lexical search text match that shares the exact same Rated Current**, checked purely on text similarity, before the Rated Current filter is even applied in `find_alternatives()`. 

```python
best_scores = []
for i in range(0, len(parts)):
    best_score = similarity_matrix[i].max()
    best_scores.append(best_score)

pd.Series(best_scores).plot(kind='hist', bins=20)
```

```python
feature_names = vectorizer.get_feature_names_out()

part_index = 0
best_match_index = similarity_matrix[part_index].argmax()

vector_a = vectors[part_index].toarray()[0]
vector_b = vectors[best_match_index].toarray()[0]

contributions = []
for i in range(0, len(feature_names)):
    word_contribution = vector_a[i] * vector_b[i]
    if word_contribution > 0:
        contributions.append((feature_names[i], word_contribution))

contributions.sort(key=lambda pair: pair[1], reverse=True)

print("Part       :", parts['ID'][part_index], "-", parts['DESCRIPTION'][part_index])
print("Best match :", parts['ID'][best_match_index], "-", parts['DESCRIPTION'][best_match_index])
print("Similarity score:", similarity_matrix[part_index][best_match_index])
print()
print("Top contributing keywords:")
for word, score in contributions[:10]:
    print(word, ":", f"{score:.4f}")
```

```python
def find_alternatives(part_id, k=5):
    query_index = None
    for i in range(0, len(parts)):
        if parts['ID'][i] == part_id:
            query_index = i
            break

    if query_index is None:
        print("Part not found (or it has no DESCRIPTION):", part_id)
        return

    query_current = parts['Rated Current (A) - cleaned'][query_index]

    candidates = []
    for i in range(0, len(parts)):
        if i == query_index:
            continue
        if pd.notna(query_current):
            candidate_current = parts['Rated Current (A) - cleaned'][i]
            if candidate_current != query_current:
                continue
        score = similarity_matrix[query_index][i]
        candidates.append((score, i))

    candidates.sort(key=lambda pair: pair[0], reverse=True)
    top_candidates = candidates[:k]

    print("Query part:", parts['ID'][query_index], "-", parts['DESCRIPTION'][query_index])
    if pd.isna(query_current):
        print("Warning: Rated Current is missing, results are text-similarity only (unverified)")
    print()
    for score, i in top_candidates:
        print(parts['ID'][i], ":", f"{score:.4f}", "-", parts['DESCRIPTION'][i])

find_alternatives('A1')
```

## Summary

**Data:** `Parts.csv` has 998 rows, 32 columns, and only 66.4% of rows have a DESCRIPTION.

Built a lexical search (TF-IDF + cosine similarity) over DESCRIPTION, with a tokenizer that keeps decimals  stay distinct. 85.7% of parts have a top text match that also shares the same Rated Current, which is strong evidence the method is picking up something real, not coincidental wording. 

FAISS was considered and intentionally skipped — the dataset is small enough that exact brute-force search is already fast, so FAISS would only add complexity.
